# Criando novas colunas

Uma das operações mais comuns em um DataFrame é criar novas colunas.

Essas colunas podem ser resultado de cálculos, concatenação de textos ou qualquer outra operação realizada sobre os dados existentes.

````python
dados = {
    "Nome": ["Ana", "Bruno", "Carlos", "Daniela"],
    "Idade": [18, 25, 32, 40],
    "Salario": [2200, 3500, 4800, 6100]
}

df = pd.DataFrame(dados)

df
````

## Criando uma coluna através de um cálculo

Podemos utilizar uma coluna existente para criar outra.

Neste exemplo vamos calcular um bônus correspondente a **10% do salário**.

````python
df["Bonus"] = df["Salario"] * 0.10

df
````

## Criando outra coluna utilizando colunas existentes

Também podemos combinar diferentes colunas para gerar novos valores.

Neste exemplo vamos calcular o salário total considerando o bônus.

````python
df["Salario_Final"] = df["Salario"] + df["Bonus"]

df
````

## Criando colunas com textos

Nem toda coluna precisa ser numérica.

Também podemos criar informações em formato de texto.

````python
df["Pais"] = "Brasil"

df
````

# Aplicando funções em colunas

Até agora utilizamos funções próprias do Pandas, como `str.strip()` e `str.title()`.

Entretanto, em muitas situações precisamos executar uma função personalizada para cada valor de uma coluna.

Para isso, utilizamos a função **`apply()`**.

O `apply()` percorre todos os valores de uma coluna e aplica uma função em cada um deles.

````python
dados = {
    "Nome":["Ana","Bruno","Carlos","Daniela"],
    "Idade":[18,25,32,40],
    "Salario":[2200,3500,4800,6100]
}

df = pd.DataFrame(dados)

df
````

## Utilizando uma função pronta

O método `apply()` também pode utilizar funções já existentes do Python.

Neste exemplo, vamos converter todos os nomes para letras maiúsculas utilizando a função `str.upper`.

````python
df["Nome"] = df["Nome"].apply(str.upper)

df
````

## Utilizando funções `lambda`

Uma função **lambda** é uma função simples, escrita em apenas uma linha.

Ela é bastante utilizada quando a transformação é pequena e será utilizada apenas uma vez.

````python
dados = {
    "Nome":["Ana","Bruno","Carlos","Daniela"],
    "Salario":[2200,3500,4800,6100]
}

df = pd.DataFrame(dados)

df
````

Neste exemplo vamos aumentar todos os salários em **10%**.

````python
df["Salario"] = df["Salario"].apply(
    lambda salario: salario * 1.10
)

df
````

Também podemos criar novas colunas utilizando funções lambda.

````python
df["Bonus"] = df["Salario"].apply(
    lambda salario: salario * 0.10
)

df
````

## Criando uma função com `def`

Quando uma transformação é maior ou será reutilizada várias vezes, podemos criar uma função utilizando `def`.

Depois, basta utilizar essa função juntamente com o `apply()`.

````python
dados = {
    "Nome":["Ana","Bruno","Carlos","Daniela"],
    "Idade":[15,22,38,67]
}

df = pd.DataFrame(dados)

df
````

Vamos criar uma função para classificar as pessoas de acordo com a idade.

````python
def classificar_idade(idade):

    if idade < 18:
        return "Menor de idade"

    elif idade < 65:
        return "Adulto"

    else:
        return "Idoso"
````

````python
df["Categoria"] = df["Idade"].apply(classificar_idade)

df
````

Observe que a função foi executada para cada valor da coluna **Idade**, gerando uma nova coluna chamada **Categoria**.

# Transformações condicionais

Em muitas situações precisamos criar novas informações com base em uma condição.

Por exemplo:

- Identificar se uma pessoa é maior de idade.
- Classificar funcionários por faixa salarial.
- Informar se um aluno foi aprovado ou reprovado.

Essas transformações podem ser feitas utilizando **if/else**, funções **lambda** ou funções do **NumPy**.

````python
dados = {
    "Nome":["Ana","Bruno","Carlos","Daniela","Eduardo"],
    "Idade":[15,18,24,16,35]
}

df = pd.DataFrame(dados)

df
````

## Utilizando `lambda` com `if` e `else`

Quando a condição é simples, podemos utilizar uma função **lambda** juntamente com um **if/else**.

A nova coluna será criada de acordo com a condição informada.

````python
df["Maior_Idade"] = df["Idade"].apply(
    lambda idade: "Sim" if idade >= 18 else "Não"
)

df
````

#### Outro exemplo

````python
dados = {
    "Funcionario":["Ana","Bruno","Carlos","Daniela"],
    "Salario":[2200,3800,5200,7600]
}

df = pd.DataFrame(dados)

df
````

````python
df["Faixa"] = df["Salario"].apply(
    lambda salario:
    "Alto"
    if salario >= 5000
    else "Normal"
)

df
````

## Utilizando np.where()

Outra forma bastante utilizada para criar colunas condicionais é através da função **`np.where()`**.

Ela recebe três parâmetros:

- condição
- valor caso a condição seja verdadeira
- valor caso a condição seja falsa

Essa função é muito utilizada em projetos de análise de dados por ser simples e eficiente.

````python
dados = {
    "Nome":["Ana","Bruno","Carlos","Daniela","Eduardo"],
    "Idade":[15,18,24,16,35]
}

df = pd.DataFrame(dados)

df
````

````python
df["Maior_Idade"] = np.where(
    df["Idade"] >= 18,
    "Sim",
    "Não"
)

df
````

Perceba que o resultado é exatamente o mesmo obtido utilizando `lambda`.

A principal diferença é que o `np.where()` trabalha diretamente sobre a coluna inteira, sem utilizar o `apply()`.

# Concatenando DataFrames

Até agora trabalhamos com apenas um DataFrame.

Entretanto, em muitos projetos é comum possuir informações separadas em diferentes tabelas.

Para unir esses DataFrames podemos utilizar a função **`pd.concat()`**.

Ela permite juntar tabelas pelas linhas ou pelas colunas.

````python
import pandas as pd

clientes1 = pd.DataFrame({
    "ID":[1,2,3],
    "Nome":["Ana","Bruno","Carlos"]
})

clientes2 = pd.DataFrame({
    "ID":[4,5],
    "Nome":["Daniela","Eduardo"]
})

print(clientes1)
print(clientes2)
````

## Concatenando pelas linhas (`axis=0`)

Quando utilizamos **`axis=0`**, as linhas de um DataFrame são adicionadas abaixo das linhas do outro.

Esse é o comportamento padrão da função `concat()`.

````python
clientes = pd.concat([clientes1, clientes2])

clientes
````

Observe que os índices foram mantidos.

Por isso aparecem índices repetidos.

Para criar uma nova numeração dos índices, podemos utilizar o parâmetro **`ignore_index=True`**.

````python
clientes = pd.concat(
    [clientes1, clientes2],
    ignore_index=True
)

clientes
````

## Concatenando pelas colunas (`axis=1`)

Também podemos unir DataFrames lado a lado.

Para isso utilizamos **`axis=1`**.

````python
nomes = pd.DataFrame({
    "Nome":["Ana","Bruno","Carlos"]
})

idades = pd.DataFrame({
    "Idade":[18,25,30]
})

print(nomes)
print(idades)
````

````python
pd.concat(
    [nomes, idades],
    axis=1
)
````

Nesse caso, as colunas foram unidas.

É importante que ambos os DataFrames possuam a mesma quantidade de linhas.

# Relacionando DataFrames com `merge()`

Enquanto o `concat()` apenas junta tabelas, o **`merge()`** relaciona informações através de uma coluna em comum.

Esse comportamento é semelhante ao **PROCV** do Excel ou aos comandos **JOIN** utilizados em bancos de dados.

````python
clientes = pd.DataFrame({
    "ID":[1,2,3,4],
    "Nome":[
        "Ana",
        "Bruno",
        "Carlos",
        "Daniela"
    ]
})

clientes
````

````python
pedidos = pd.DataFrame({
    "ID":[1,2,2,4],
    "Produto":[
        "Mouse",
        "Notebook",
        "Monitor",
        "Teclado"
    ]
})

pedidos
````

## Relacionando os DataFrames

Observe que ambos possuem a coluna **ID**.

Essa coluna será utilizada para relacionar os registros.

````python
resultado = pd.merge(
    clientes,
    pedidos,
    on="ID"
)

resultado
````

A coluna **ID** é chamada de **chave de relacionamento**, pois é utilizada para localizar informações correspondentes nos dois DataFrames.

## `how="inner"`

O tipo **Inner Join** retorna apenas os registros existentes nos dois DataFrames.

````python
pd.merge(
    clientes,
    pedidos,
    on="ID",
    how="inner"
)
````

## `how="left"`

Retorna todos os registros do DataFrame da esquerda.

Caso não exista correspondência, o valor será preenchido com **NaN**.

````python
pd.merge(
    clientes,
    pedidos,
    on="ID",
    how="left"
)
````

## `how="right"`

Retorna todos os registros do DataFrame da direita.

````python
pd.merge(
    clientes,
    pedidos,
    on="ID",
    how="right"
)
````

## `how="outer"`

Retorna todos os registros dos dois DataFrames.

Quando não houver correspondência, os valores ausentes serão preenchidos com **NaN**.

````python
pd.merge(
    clientes,
    pedidos,
    on="ID",
    how="outer"
)
````

# Concat x Merge

Embora ambos sejam utilizados para unir informações, eles possuem objetivos diferentes.

| Concat | Merge |
|---------|--------|
| Junta DataFrames | Relaciona DataFrames |
| Não precisa de uma chave | Precisa de uma chave em comum |
| Pode unir linhas ou colunas | Une registros correspondentes |
| Utiliza `axis=0` ou `axis=1` | Utiliza `on` e `how` |
